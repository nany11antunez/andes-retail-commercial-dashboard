### Diseño y planificación

Para el diseño del Dashboard 1 (Overview Ejecutivo), la prioridad visual es la claridad, la síntesis y la toma de decisiones rápida. La vista se estructura con un encabezado superior de tarjetas de KPI para dar contexto de salud financiera inmediata, seguido de un cuerpo analítico principal centrado en la tendencia temporal interanual y desgloses macro por geografía y categoría de producto.

#### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?
KPIs:
- Ingresos Totales: Muestra el volumen monetario bruto generado por el negocio. Es la métrica principal de escala comercial para evaluar el crecimiento general entre 2024 y 2025.  
- Ganancia Bruta: Permite medir la rentabilidad real del negocio tras descontar los costos operacionales. Un aumento en ingresos debe respaldarse con sostenibilidad en la ganancia.
- Unidades Vendidas: Refleja la demanda física y el volumen transaccional puro. Permite entender si la variación en ingresos se debe a mayor cantidad de ventas o a cambios en precios.

Gráficos:

- Gráfico de tiempo: Es el estándar visual más efectivo para representar variables continuas a lo largo del tiempo. Al colocar los meses (Enero a Diciembre) en el eje X y comparar la línea de 2024 contra la de 2025 en un mismo marco de referencia, se detectan de inmediato tendencias de crecimiento, estacionalidades (picos de demanda en estaciones clave) y caídas comerciales sin saturar la vista.

- Gráfico comparación geográfica: Las barras verticales facilitan la lectura clara de los nombres de los países (Perú, Chile, Colombia) y regiones sin sobrecargar el espacio visual. Ordenar las barras de mayor a menor según los Ingresos permite al directivo jerarquizar mercados en un segundo, mientras que aplicar un degradado de color según la Ganancia Bruta expone qué país aporta mayor rentabilidad neta y no solo volumen bruto.

- Gráfico comparación clientes/productos: Permite comparar de forma directa e intuitiva el volumen total de ingresos desglosado por `Categoría_Producto` en el eje horizontal, mostrando barras verticales contiguas para cada `Segmento_Cliente` (*Estándar* vs. *Premium*). La comparación de alturas facilita una lectura ejecutiva limpia sobre qué perfil de cliente domina en cada línea de producto y en cuáles se concentran los mayores ingresos del negocio.

Descripción de la jerarquía visual:
- Título: Ubicado en la parte superior izquierda con la tipografía de mayor tamaño y peso, para contextualizar el tablero comercial de Andes Retail Group.
- Banners de KPIs:Tres tarjetas horizontales estratégicamente ubicadas (Ingresos Totales, Ganancia Bruta y Unidades Vendidas) con tipografía destacada para auditar la salud general del negocio en menos de 3 segundos.
- Gráfico Principal Evolución Temporal de Ingresos (Líneas 2024 vs 2025): Es el elemento visual de mayor área, posicionando la historia principal del negocio como el foco de atención al centro, acompañado de sus leyendas de color para el año actual vs. previo
Gráficos Secundarios:
- Ingresos por País (Barras verticales): Posicionado abajo a la izquierda para establecer el desglose geográfico de mayor a menor ingreso con su ganancia bruta de cada uno de los paises,
- Distribución de Ventas por Categoría y Perfil de Cliente (Barras Agrupadas Horizontales): Desglose posicionado a la derecha que compara el volumen de ingresos generado por los clientes Estándar vs. Premium en cada línea de producto.

Filtros:
- Año: Filtro global desplegable para seleccionar Todos, 2024 o 2025, actualizando métricas y gráficos en tiempo real.
- País: Filtro secundario para aislar los datos por mercado geográfico.

#### Vista detalle - ¿Por qué se han caido/subido/mantenido los ingresos?

Justificación: Muestra la composición interna de la cartera de clientes de cada país para diagnosticar si la baja rentabilidad de Colombia se debe a una menor penetración del segmento Premium.

#### Vista General (Overview)

**S (Situación):**  
Andes Retail Group alcanzó un volumen total de ingresos de 5.531.994, generando una ganancia bruta de $1.941.749 (margen del 35,1%) a través de 57.601 unidades vendidas entre 2024 y 2025.

**C (Complicación):**  
El crecimiento temporal se encuentra estancado (la trayectoria de ingresos de 2025 réplica de cerca a la de 2024). Se observa una fuerte estacionalidad con una caída continua que inicia en febrero y se profundiza abruptamente en los meses centrales del año (junio a agosto), donde los ingresos tocan su nivel mínimo antes de recuperar en el último cuatrimestre y existe una disparidad de margen geográfico: Colombia registra los ingresos más bajos (1.345.747) y el nivel más bajo de ganancia bruta (470.321), contrastando con la alta rentabilidad de Perú y Chile ($760.633).

**Q (Pregunta):** ¿Cómo podemos mitigar el valle de ventas del periodo junio-agosto y mejorar la rentabilidad de Colombia sin descuidar a los mercados principales?  

**A (Respuesta):** Lanzar campañas de reactivación a mitad de año apalancadas en el segmento Premium (que domina en todas las categorías) y revisar la estrategia comercial en Colombia para alinearlo a los márgenes de Perú y Chile.

#### Vista Detalle

**S (Situación):**  
Andes Retail Group opera en el mercado sudamericano (Perú, Chile y Colombia) manteniendo un porcentaje de margen bruto altamente estable, ubicado de forma constante entre el 34,67% y el 35,62% en todas sus regiones y categorías de producto (Deportes, Electrónica, Hogar y Ropa).

**C (Complicación):**  
Se observa una fuerte vulnerabilidad estacional con una caída drástica de la facturación en los meses de invierno del hemisferio sur (junio a agosto, alcanzando un mínimo de 45.172 en julio frente al pico de 383.572 en enero). Asimismo, Colombia presenta un menor peso en el segmento de mayor valor, registrando solo un 44,28% de clientes Premium en comparación con Perú (47,72%) y Chile (47,95%).

**Q (Pregunta):**  
¿Qué acciones comerciales y operativas se deben implementar para mitigar la desaceleración de ventas a mitad de año y acelerar la participación de clientes Premium en Colombia sin alterar el margen bruto actual?

**A (Respuesta):**  
Se recomiendan dos líneas de acción estratégicas que protegen la rentabilidad unitaria del ~35%:

- Captación y Conversión Premium en Colombia: Reasignar el presupuesto publicitario existente hacia audiencias de mayor poder adquisitivo para optimizar la adquisición en el origen. A la vez, incentivar el up-selling en la base actual mediante programas de fidelización basados en beneficios no monetarios (como soporte prioritario o acceso anticipado a lanzamientos) que no sacrifiquen margen.
- Sostenimiento Estacional con Venta Cruzada (Cross-selling): Ejecutar campañas de venta cruzada entre categorías complementarias (como Deportes y Electrónica) durante el trimestre junio-agosto. Esto eleva el ticket promedio sin recurrir a descuentos sobre precio, sosteniendo el volumen de facturación y manteniendo el margen bruto intacto.
